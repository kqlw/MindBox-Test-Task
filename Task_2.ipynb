```python
from os import terminal_size
class GeometricShapes():
  def __init__(self):
    self._pi = 3.14
  def square(self):
    pass


class Circle(GeometricShapes):
  def __init__(self, r):
    super().__init__()

    if r < 0:
      raise ValueError("Радиус не может быть < 0")
    else:
      self.__r = r

  @property
  def radius(self, r):
    return self.__r

  @radius.setter
  def radius(self, r):
    if r < 0:
      raise ValueError("Радиус не может быть < 0")
    else:
      self.__r = r

  def square(self):
    return  self._pi * (self.__r ** 2)


class Triangle(GeometricShapes):
  def __init__(self, a, b, c):
    super().__init__()

    if self.__isTriangle(a, b, c):
      self.__a = a
      self.__b = b
      self.__c = c
    else:
      raise ValueError("Треугольник существует тогда и только тогда, когда сумма любых двух его сторон больше третьей стороны!")

  def __isTriangle(self, a, b, c):
    if ((a + b ) > c) and ((a + c ) > b) and ((c + b ) > a) :
        return True
    else:
        return False
  def isRightAngledTriangle(self):
    if ((self.__a**2 + self.__b**2 ) == self.__c**2) or ((self.__a**2 + self.__c**2 ) == self.__b**2) or ((self.__c**2 + self.__b**2 ) == self.__a**2):
        return True
    else:
        return False

  def square(self):
    p = (self.__a + self.__b + self.__c) / 2
    s = (p * ((p - self.__a)*(p - self.__b)*(p - self.__c)) ) ** 0.5

    return s



```

```python
c = Circle(10)

c.square()
```
